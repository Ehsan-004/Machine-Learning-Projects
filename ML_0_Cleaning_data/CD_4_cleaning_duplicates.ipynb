If there are duplicate data in your dataset you can use `duplicated()` method.\
In our example, there is a duplicate value in row 3, 4.\
By the way removing duplicates can't effect on your result.

```python
import pandas as pd

df = {
    "duration": [
        20,
        40,
        50,
        50,
        30,
        60],
    "pulse": [
        100,
        140,
        120,
        120,
        110,
        100]
}

df = pd.DataFrame(df)
```

You can check duplicates with `duplicated()` method.

```python
df.duplicated()
```
