## `fillna()` ##

`fillna(arg)` would fill all empty cells with the input arg.

```python
ndf2 = df.copy()
# ndf2['Pulse'].fillna(103, inplace=True)
ndf2.fillna({'Pulse': 103}, inplace=True)
# ndf2['Duration'].fillna(45, inplace=True)
ndf2.fillna({'Duration': 45}, inplace=True)
ndf2.shape
```

If you want to use `fillna()`, you can set mean, mode or median value of filed as filling value.\
In some conditions, it cannot  make any change in final result.

```python
x = df['Pulse'].mean()
x = df['Pulse'].median()
x = df['Pulse'].mode()

ndf3 = df.copy()
ndf3.fillna({'Pulse': x}, inplace=True)
ndf3.shape
```
