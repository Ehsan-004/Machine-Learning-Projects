<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
    در ابتدا نیاز است فایل‌های مجموعه‌داده را بخوانید. نمونه‌های آموزشی در فایل <code>train.csv</code> و نمونه‌های آزمون که باید مقدار متغیر هدف آن‌ها را پیش‌بینی کنید در فایل <code>test.csv</code> ذخیره شده‌اند. اگر لازم دانستید می‌توانید به دلخواه خود بخشی از مجموعه‌ی آموزش را به عنوان مجموعه‌ی اعتبارسنجی نیز جدا کنید.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
from sklearn.metrics import f1_score
```

```python
df = pd.read_csv("./train.csv")
df["author"].value_counts()
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
پیش‌پردازش و مهندسی ویژگی
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
        در این سوال شما می‌توانید از هر تکنیک پیش‌پردازش/مهندسی ویژگی دلخواهتان، استفاده کنید.
    <br>
    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت؛ بنابراین هر چه پیش‌پردازش/مهندسی ویژگی بهتری انجام دهید تا دقت مدل بهبود پیدا کند، امتیاز بیشتری از این سوال کسب خواهید کرد.

</font>
</p>

```python
df['author'] = preprocessing.LabelEncoder().fit(df['author'].unique()).transform(df['author'])

x = df.copy()
x.drop('author', axis=1, inplace=True)
y = df['author']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=5)
print("training data: ", x_train.shape, y_train.shape)
print("testing data: ", x_test.shape, y_test.shape)
```

<h2 dir=rtl align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
آموزش مدل
</font>
</h2>

<p dir=rtl style="direction: rtl;text-align: justify;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazir" size=3>
    حال که داده را پاکسازی کرده و احتمالا ویژگی‌هایی را به آن افزوده یا از آن حذف کرده‌اید، وقت آن است که مدلی آموزش دهید که بتواند متغیر هدف این مسئله را پیش‌بینی کند.
</font>
</p>

```python
from sklearn import svm

clf_svm = svm.SVC(C=2, kernel='rbf')
clf_svm.fit(x_train, y_train)
```

```python
p_svm = clf_svm.predict(x_test)
```

```python
from sklearn.linear_model import RidgeClassifier
clf_rc = RidgeClassifier()
clf_rc.fit(x_train, y_train)
```

```python
p_rc = clf_rc.predict(x_test)
```

```python
from sklearn.linear_model import SGDClassifier
clf_sgd = SGDClassifier()
clf_sgd.fit(x_train, y_train)
```

```python
p_sgd = clf_sgd.predict(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import f1_score

# ks = 100
# for i in range(1, ks):
#     clf_kn = KNeighborsClassifier(n_neighbors=i)
#     clf_kn.fit(x_train, y_train)
#     p = clf_kn.predict(x_test)
#     print(f1_score(y_test, p, average='macro'), i)

# best K is 2
clf_kn = KNeighborsClassifier(n_neighbors=2)
clf_kn.fit(x_train, y_train)
```

```python
p_kn = clf_kn.predict(x_test)
```

```python
from sklearn.neighbors import RadiusNeighborsClassifier
als = ['ball_tree', 'kd_tree', 'brute']
for al in als:
    x_train_new = x_train.to_numpy().reshape(len(x_train)-1, 1)
    clf_rn = RadiusNeighborsClassifier(radius=1.0)
    clf_rn.fit(x_train_new, y_train)
    p = clf_rn.predict(x_test)
    print(f1_score(y_test, p, average='macro'), al)
```
