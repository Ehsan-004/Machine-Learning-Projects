    در ابتدا نیاز است فایل‌های مجموعه‌داده را بخوانید. نمونه‌های آموزشی در فایل <code>train.csv</code> و نمونه‌های آزمون که باید مقدار متغیر هدف آن‌ها را پیش‌بینی کنید در فایل <code>test.csv</code> ذخیره شده‌اند. اگر لازم دانستید می‌توانید به دلخواه خود بخشی از مجموعه‌ی آموزش را به عنوان مجموعه‌ی اعتبارسنجی نیز جدا کنید.

```python
test_data = pd.read_csv("../data/test.csv")
df = pd.read_csv("../data/train.csv") # my train_data
```

<h2 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
پیش‌پردازش و مهندسی ویژگی
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
    در این سوال شما می‌توانید از هر تکنیک پیش‌پردازش یا مهندسی ویژگی، استفاده کنید.
    <br>
    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت.
</font>
</p>

```python
# Convert non numeric values to numeric values:
df['status'] = preprocessing.LabelEncoder().fit(df['status'].unique()).transform(df['status'])
df['allocation'] = preprocessing.LabelEncoder().fit(df['allocation'].unique()).transform(df['allocation'])
df['city.id'] = preprocessing.LabelEncoder().fit(df['city.id'].unique()).transform(df['city.id'])
df['province.id'] = preprocessing.LabelEncoder().fit(df['province.id'].unique()).transform(df['province.id'])
df['is_new'] = preprocessing.LabelEncoder().fit(df['is_new'].unique()).transform(df['is_new'])
df['is_clean'] = preprocessing.LabelEncoder().fit(df['is_clean'].unique()).transform(df['is_clean'])
df['is_plus'] = preprocessing.LabelEncoder().fit(df['is_plus'].unique()).transform(df['is_plus'])
df['is_instant'] = preprocessing.LabelEncoder().fit(df['is_instant'].unique()).transform(df['is_instant'])
# delete stays_max field and replace with has_max_limit instead
df.fillna({'stays_max': 0}, inplace=True)
has_sm = list(df['stays_max'] == 0)
df['has_max_limit'] = preprocessing.LabelEncoder().fit([True, False]).transform(has_sm)
df.drop(labels=['stays_max'], axis=1);

df['types'] = df['types'].str.replace('[','').str.replace(']','').str.replace(' ','').str.replace("'",'').apply(lambda x: x.strip(','))
df['regions'] = df['regions'].str.replace('[','').str.replace(']','').str.replace(' ','').str.replace("'",'').apply(lambda x: x.strip(','))
df['rules'] = df['rules'].str.replace('[','').str.replace(']','').str.replace(' ','').str.replace("'",'').apply(lambda x: x.strip(','))


ndf = df[['status', 'allocation', 'floor_area', 'land_area', 'floors_count',
       'bedrooms', 'guest_number',
       'max_guest_number', 'stays_min', 'has_max_limit', 'entrance_time_min',
       'entrance_time_max', 'leaving_time', 'min_price',
       'is_clean', 'is_new', 'is_instant', 'is_plus', 'geo.lat', 'geo.lng',
       'province.id', 'city.id',
       'ratings.count', 'ratings.total', 'ratings.cleanliness',
       'ratings.location', 'ratings.checkin', 'ratings.value', 'units_count',
       'success_books']]

for index, row in df.iterrows():
    for ty in row['types'].split(','):
        ndf.at[index, ty] = 1

for index, row in df.iterrows():
    for ty in row['regions'].split(','):
        ndf.at[index, ty] = 1

for index, row in df.iterrows():
    for ty in row['rules'].split(','):
        ndf.at[index, ty] = 1
# ndf.drop(columns=[' '], axis=1, inplace=True);
ndf.fillna(0, inplace=True)

x = ndf[['status', 'allocation', 'floor_area', 'land_area', 'floors_count',
         'bedrooms', 'guest_number', 'max_guest_number', 'stays_min',
        'has_max_limit', 'entrance_time_min', 'entrance_time_max',
        'leaving_time', 'is_clean', 'is_new', 'is_instant',
        'is_plus', 'geo.lat', 'geo.lng', 'province.id', 'city.id',
        'ratings.count', 'ratings.total', 'ratings.cleanliness',
        'ratings.location', 'ratings.checkin', 'ratings.value', 'units_count',
        'success_books', 'suite', 'villa', 'boutiquehotel', 'ecolog',
        'apartment', 'cottage', 'swiss_cottage', 'wooden_cottage',
        'apartmenthotel', 'caravansera', 'ruralhome', 'hostel', 'motel',
        'rural', 'mountain', 'urban', 'suburb', 'forest', 'desert', 'beach', '',
        'pet', 'smoke', 'party', 'unmarried']].values

y = ndf['min_price'].values

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=5)
```

<h2 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
مدل‌سازی
</font>
</h2>

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
پس از پاکسازی داده‌ها و احتمالا حذف بعضی ستون‌ها یا افزودن ویژگی‌های جدید، حال وقت آن است که مطابق سلیقه‌ی خود، مدل مناسب را با استفاده از داده‌های پیش‌پردازش شده آموزش دهید.
</font>
</p>

```python
model = GradientBoostingRegressor(random_state=2).fit(x_test, y_test)
```

<font face="vazir" color="#0099cc">

<p dir=rtl style="direction: rtl; text-align: justify; line-height:200%; font-family:vazir; font-size:medium">
<font face="vazir" size=3>
    در این سوال از معیار  R2 Score جهت ارزیابی نتایج مدل شما استفاده خواهد شد.
    در سامانه داوری هم از این معیار برای نمره‌دهی استفاده شده است.
    <br>
    پیشنهاد می‌شود با توجه به این معیار، عملکرد مدل خود را بر روی مجموعه‌ی آموزش یا اعتبارسنجی ارزیابی کنید.

```python
prediction = model.predict(x_test)

model_r2_score = r2_score(y_test, prediction)
print(model_r2_score)
```
